# Task 3: Bivariate Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv(r"C:\Users\Sadhana sri\Downloads\airlines_flights_data.csv")
```

```python
# Set plot style
sns.set(style="whitegrid")
```

# 1. Airline vs Price

```python
plt.figure(figsize=(10,5))
sns.boxplot(x='airline', y='price', data=df)
plt.xticks(rotation=45)
plt.title("Airline vs Ticket Price")
plt.show()
```

```python
# 2. Class vs Price
plt.figure(figsize=(6,5))
sns.boxplot(x='class', y='price', data=df)
plt.title("Travel Class vs Ticket Price")
plt.show()
```

```python
# 3. Stops vs Price
plt.figure(figsize=(6,5))
sns.boxplot(x='stops', y='price', data=df)
plt.title("Stops vs Ticket Price")
plt.show()
```

```python
# 4. Duration vs Price
plt.figure(figsize=(8,5))
sns.scatterplot(x='duration', y='price', data=df)
plt.title("Flight Duration vs Ticket Price")
plt.show()
```

```python
# 5. Days Left vs Price
plt.figure(figsize=(8,5))
sns.scatterplot(x='days_left', y='price', data=df)
plt.title("Days Left vs Ticket Price")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Sadhana sri\Downloads\airlines_flights_data.csv")
```

```python
# Encode categorical variables
df["airline"] = df["airline"].astype("category").cat.codes
df["class"] = df["class"].astype("category").cat.codes
df["stops"] = df["stops"].astype("category").cat.codes
```

```python
# Select numerical columns
data = df.select_dtypes(include="number")
```

```python
# Correlation
corr = data.corr()
```

```python
print(corr)
```

```python
# Heatmap
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Feature 1: Price per day left
df["price_per_day"] = df["price"] / (df["days_left"] + 1)

```

```python
# Feature 2: Duration category
df["long_flight"] = df["duration"] > df["duration"].mean()
```
